# Exploratory Data Analysis (EDA)

Before we jump into training a classifier, we need to understand our text data and figure out how we are going to create our training labels. 

```python
import pandas as pd
import plotly.express as px
```

Load the raw data

```python
df = pd.read_csv('../data/raw/pickme_reviews_with_sentiment.csv')
```

Filter for English only and reset index

```python
df_en = df[df['language'] == 'English'].copy().reset_index(drop=True)
```

Create a text length feature (useful for modeling later)

```python
df_en['review_length'] = df_en['review_text'].str.len()
```

Display basic data checks

```python
print(f"Total English Reviews: {len(df_en)}")
print("\nSentiment Distribution:")
print(df_en['sentiment'].value_counts())
```

Plotly Visualizations

```python
# A: Sentiment Count
df_counts = df_en['sentiment'].value_counts().reset_index(name='count').rename(columns={'index': 'sentiment'})
fig_sentiment = px.bar(df_counts, 
                       x='sentiment', y='count', 
                       title='Sentiment Distribution (English Reviews)',
                       labels={'sentiment': 'Sentiment', 'count': 'Count'})
fig_sentiment.show()
```

```python
# B: Review Length by Sentiment (Box Plot to see outliers)
fig_length = px.box(df_en, x='sentiment', y='review_length',
                    title='Review Length Distribution by Sentiment',
                    color='sentiment')
fig_length.show()
```

Negative reviews tend to be longer (more detailed complaints) than positive ones
